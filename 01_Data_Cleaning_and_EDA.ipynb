# **Data Cleaning & Preprocessing**

**Make a Copy of the Dataset**

```python
df_clean = df.copy()
```

**Remove Duplicate Rows**

```python
print("Duplicates Before:", df_clean.duplicated().sum())

df_clean.drop_duplicates(inplace=True)

print("Duplicates After:", df_clean.duplicated().sum())
```

Remove Invalid Ratings

```python
df_clean = df_clean[df_clean["Rating"] <= 5]
```

**Fill Missing Ratings**

```python
df_clean["Rating"].fillna(df_clean["Rating"].median(), inplace=True)
```

**Convert Reviews to Numeric**

```python
df_clean["Reviews"] = pd.to_numeric(df_clean["Reviews"], errors="coerce")
```

**Clean Installs Column**

```python
df_clean["Installs"] = (
    df_clean["Installs"]
    .str.replace("+", "", regex=False)
    .str.replace(",", "", regex=False)
)

df_clean["Installs"] = pd.to_numeric(df_clean["Installs"], errors="coerce")
```

**Clean Price Column**

```python
df_clean["Price"] = (
    df_clean["Price"]
    .str.replace("$", "", regex=False)
)

df_clean["Price"] = pd.to_numeric(df_clean["Price"], errors="coerce")
```

**Convert** Size to MB

```python
def convert_size(size):
    if size == "Varies with device":
        return np.nan

    if "M" in size:
        return float(size.replace("M",""))

    elif "k" in size:
        return float(size.replace("k","")) / 1024

    else:
        return np.nan

df_clean["Size"] = df_clean["Size"].apply(convert_size)
```

**Convert Last Updated to Date**

```python
df_clean["Last Updated"] = pd.to_datetime(df_clean["Last Updated"])
```

**Fill Remaining Missing Values**

```python
df_clean["Size"].fillna(df_clean["Size"].median(), inplace=True)

df_clean["Type"].fillna(df_clean["Type"].mode()[0], inplace=True)

df_clean["Current Ver"].fillna("Unknown", inplace=True)

df_clean["Android Ver"].fillna("Unknown", inplace=True)
```

**Check Missing Values Again**

```python
df_clean.isnull().sum()
```

**Final Dataset Information**

```python
df_clean.info()
```

# Feature Engineering

**App Age**

```python

df_clean["App Age"] = 2024 - df_clean["Last Updated"].dt.year
```

**Rating Category**

```python
df_clean["Rating Category"] = pd.cut(
    df_clean["Rating"],
    bins=[0, 2, 3.5, 4.5, 5],
    labels=["Poor", "Average", "Good", "Excellent"]
)
```

**Install Category**

```python
df_clean["Install Category"] = pd.cut(
    df_clean["Installs"],
    bins=[0, 1000, 10000, 100000, 1000000, np.inf],
    labels=["<1K", "1K-10K", "10K-100K", "100K-1M", ">1M"]
)
```

**Size Category**

```python
df_clean["Size Category"] = pd.cut(
    df_clean["Size"],
    bins=[0, 10, 30, 60, np.inf],
    labels=["Small", "Medium", "Large", "Very Large"]
)
```

**Premium App**

```python
df_clean["Premium App"] = np.where(df_clean["Type"] == "Paid", 1, 0)
```

**Revenue Estimate**

```python
df_clean["Estimated Revenue"] = (
    df_clean["Price"] * df_clean["Installs"]
)
```

**Reviews per Install**

```python
df_clean["Reviews per Install"] = (
    df_clean["Reviews"] / df_clean["Installs"]
)
```

**Top Rated App**

```python
df_clean["Top Rated"] = np.where(
    df_clean["Rating"] >= 4.5,
    "Yes",
    "No"
)
```

**App Size (MB)**

```python
df_clean.rename(columns={"Size": "Size (MB)"}, inplace=True)
```

**IQR (Interquartile Range)**

```python
Q1 = df_clean["Rating"].quantile(0.25)
Q3 = df_clean["Rating"].quantile(0.75)

IQR = Q3 - Q1

print("Q1 :", Q1)
print("Q3 :", Q3)
print("IQR:", IQR)
```

**Outlier Detection (Rating)**

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean["Rating"] < lower) |
    (df_clean["Rating"] > upper)
]

print("Number of Outliers:", len(outliers))
```
